# **Lab Report 1: GCD (Euclidean Algorithm)**


## **Theory**


The Euclidean algorithm is based on a simple but powerful mathematical principle,

For any integers $a$ and $b$ with $b≠0$,

$\hspace{2cm}gcd(a, b) = gcd(b, a\:mod\:b)$

Where $a\:mod\:b = a - b\: \cdot \:[a/b]$

## **Algorithm (Pseudocode)**


```
GCD(a, b):
    if a == 0:
        return b
    else if b == 0:
        return a
    else:
        while b != 0:
            r = a % b
            a = b
            b = r

        return a;
```


## **Code & Output**

````python
def GCD(a, b):
    a, b = abs(a), abs(b)
    
    if a == 0:
        return b
    elif b == 0:
        return a
    
    while b != 0:
        a, b = b, a % b
    
    return a


if __name__ == "__main__":
    a = int(input("Enter a: "))
    b = int(input("Enter b: "))

    result = GCD(a, b)
    
    print(f"The greatest common divisor of {a} and {b} is: {result}")
````

## **Complexity Analysis**


#### **Time Complexity**


1. **Best Case: $O(1)$** <br>
The best case occurs when the algorithm terminates in a single step i.e., either $b = 0$ or $a \: mod \: b = 0$ <br>
Thus, $B(n) = O(1)$

2. **Average Case: $O(log\:n)$** <br>
Each step replaces $(a,b)→(b,a\:mod\:b)$ <br>
The remainder is significantly smaller than the previous number which leads to a logarithmic number of steps relative to the size of the input. <br>
Thus, $A(n) = O(log\:n)$

3. **Worst Case: $O(log(min(a, b)))$** <br>
The worst case occurs when the reduction is as slow as possible. This happens when a and b are consecutive Fibonacci numbers. <br>
Thus, $W(n) = O(log(min(a, b)))$


#### **Space Complexity** <br>


Only a few variables used $(a, b, r)$ for the iterative method. <br>
Thus, $S(n) = O(1)$

## **Conclusion**


The iterative GCD algorithm is highly efficient with logarithmic time and constant space, making it ideal for production use—even on massive integers. It outperforms naive factorization methods $O(\sqrt{n})$ and exemplifies algorithmic elegance: simple, provably optimal, and foundational for cryptography/modular arithmetic.
