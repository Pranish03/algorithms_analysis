# **Lab Report 10: 0/1 Knapsack Problem**


## **Theory**


The 0/1 Knapsack Problem is a classic optimization problem: given n items, each with a weight and a value, determine which items to include in a knapsack of capacity W to maximize the total value, with each item either fully included $(1)$ or excluded $(0)$.

The Dynamic Programming approach builds a 2D table dp[i][w] representing the maximum value achievable using the first i items with knapsack capacity w. The optimal substructure ensures that each sub-problem is solved once, avoiding the exponential time of brute-force enumeration. This problem is NP-hard in general, but the DP solution runs in pseudo-polynomial time $O(nW)$.

## **Algorithm (Pseudocode)**


```
ALGORITHM Knapsack01_DP(values, weights, n, W)
    Input: n items with values[], weights[], capacity W
    Output: Maximum value achievable

    Create dp[0..n][0..W] initialized to 0
    FOR i <- 1 TO n DO
        FOR w <- 0 TO W DO
        IF weights[i-1] <= w THEN
            dp[i][w] <- MAX(dp[i - 1][w],
                        dp[i - 1][w - weights[i - 1]] + values[i - 1])
        ELSE
            dp[i][w] <- dp[i - 1][w]
        END IF
        END FOR
    END FOR
    RETURN dp[n][W]
```


## **Code & Output**

````python
def knapsack_01(values, weights, capacity):
    n = len(values)
    dp = [[0] * (capacity + 1) for _ in range(n + 1)]

    for i in range(1, n + 1):
        for w in range(capacity + 1):
            if weights[i - 1] <= w:
                dp[i][w] = max(
                    dp[i - 1][w], dp[i - 1][w - weights[i - 1]] + values[i - 1]
                )
            else:
                dp[i][w] = dp[i - 1][w]

    selected, w = [], capacity
    for i in range(n, 0, -1):
        if dp[i][w] != dp[i - 1][w]:
            selected.append(i - 1)
            w -= weights[i - 1]

    return dp[n][capacity], selected[::-1]


if __name__ == "__main__":
    values = [60, 100, 120]
    weights = [10, 20, 30]
    capacity = 50

    max_val, items = knapsack_01(values, weights, capacity)
    print(f"Items (index) : {items}")
    print(f"Values chosen : {[values[i] for i in items]}")
    print(f"Weights chosen: {[weights[i] for i in items]}")
    print(f"Maximum Value : {max_val}")
````

## **Complexity Analysis**


#### **Time Complexity**


We fill a DP table of size $n × W$, where each state takes constant time. <br>
Here,<br>
&emsp;&emsp;$n$ = number of items <br>
&emsp;&emsp;$W$ = knapsack capasity

Thus, $T(n) = O(nW)$


#### **Space Complexity** <br>


1. **Standard DP table approach** <br>
In this approach, we store a 2D DP table of size $n × W$. <br>
Thus, $S(n) = O(nW)$

2. **Optimized 1D DP approach**
In this approach, we use a single array of size $W$ and overwrite previous states. <br>
Thus $S(n) = O(W)$


## **Conclusion**


The Dynamic Programming approach to the 0/1 Knapsack problem has pseudo-polynomial time complexity $O(nW)$, which is efficient when $W$ is small. The space can be optimized to $O(W)$ using a rolling 1D array. While NP-hard in general (exponential in W's bit length), this DP solution is practical for reasonable capacities and is fundamental in resource allocation, budget optimization, and cryptography (subset-sum problems).
