# **Lab Report 2: Fibonacci Sequence**


## **Theory**


The Fibonacci sequence is defined where each number is the sum of the two preceding ones: $F(0)=0, F(1)=1, F(n)=F(n-1)+F(n-2)$. While a naive recursive implementation has exponential time complexity $O(2^n)$, the iterative approach computes values in linear time by tracking only the last two values.

The iterative method is preferred in practice because it avoids the stack overhead and redundant recomputation associated with naive recursion. It is commonly used in algorithm design to illustrate the power of dynamic programming / memoization principles.


## **Algorithm (Pseudocode)**


```
ALGORITHM Fibonacci_Iterative(n)
    Input: Non-negative integer n
    Output: nth Fibonacci number

    IF n == 0 THEN RETURN 0
    IF n == 1 THEN RETURN 1
    a <- 0
    b <- 1
    FOR i <- 2 TO n DO
        temp <- a + b
        a <- b
        b <- temp
    END FOR
    RETURN b
```


## **Code & Output**

````python
def fibonacci_iterative(n):
    if n == 0:
        return 0
    if n == 1:
        return 1

    a, b = 0, 1
    for _ in range(2, n + 1):
        a, b = b, a + b

    return b


if __name__ == "__main__":
    print("n   : " + " ".join(f"{i:>3}" for i in range(10)))
    print("F(n): " + " ".join(f"{fibonacci_iterative(i):>3}" for i in range(10)))
````

## **Complexity Analysis**


#### **Time Complexity**


The function has a single for loop that runs from $2$ to $n$, iterating $n-2$ times. Combined with the constant-time operations before it. <br>
Thus, $T(n)=O(n)$.


#### **Space Complexity** <br>


Only two variables $(n1, n2)$ are used regardless of $n$ — no array or recursion stack is needed. This is the optimal iterative approach. <br>
Thus, $S(n) = O(1)$

## **Conclusion**


The iterative Fibonacci algorithm achieves $O(n)$ time complexity with only $O(1)$ space, a vast improvement over the naive recursive $O(2^n)$ approach. By maintaining only two variables at any point, memory is used efficiently. This approach demonstrates how iterative solutions frequently outperform their recursive counterparts in both time and space efficiency.
