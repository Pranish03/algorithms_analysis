# **Lab Report 4: Bubble Sort**


## **Theory**


Bubble Sort is a comparison-based sorting algorithm that repeatedly steps through the list, compares adjacent elements, and swaps them if they are in the wrong order. With each full pass, the largest unsorted element 'bubbles up' to its correct position at the end of the array.

Although Bubble Sort is not efficient for large datasets due to its quadratic time complexity, it serves as an important educational tool for understanding sorting algorithms. An optimized version stops early if no swaps occur during a pass, making it $O(n)$ on already-sorted input.

## **Algorithm (Pseudocode)**


```
ALGORTIHM BubbleSort(arr, n)
    Input: Array arr of n elements
    Output: Sorted array in ascending order

    FOR i <- 0 TO n - 2 DO
        swapped <- FALSE
        FOR j <- 0 TO n - 2 - i DO
            IF arr[j] > arr[j + 1] THEN
                SWAP arr[j] and arr[j + 1]
                swapped <- TRUE
            END IF
        END FOR
        IF swapped == FALSE THEN
            BREAK
        ENDIF
    END FOR
    RETURN arr
```


## **Code & Output**

````python
def bubble_sort(arr, n):
    for i in range(n - 1):
        swapped = False

        for j in range(n - i - 1):
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
                swapped = True

        if not swapped:
            break

    return arr


if __name__ == "__main__":
    datasets = {
        "Random Array": [6, 3, 1, 7, 4, 2, 5],
        "Sorted Array": [1, 2, 3, 4, 5, 6, 7],
        "Reversed Array": [7, 6, 5, 4, 3, 2, 1],
    }

    for name, data in datasets.items():
        result = bubble_sort(data[:], len(data))
        print(f"{name:<15}: {data} -> {result}")
````

## **Complexity Analysis**


#### **Time Complexity**


1. **Best Case: $O(n)$** <br>
The best case occurs when the array is already sorted (with early-exit flag). <br>
Thus, $B(n) = O(n)$

2. **Average Case: $O(n^2)$** <br>
The average case occurs when the items in the array is in random order. <br>
Thus, $A(n) = O(n^2)$

3. **Worst Case: $O(n^2)$** <br>
The worst case occurs when the array is reversly sorted. <br>
Thus, $W(n) = O(n^2)$


#### **Space Complexity** <br>


This algorithm does not use any extra memory that grows with input size $n$. <br>
Thus, $S(n) = O(1)$


## **Conclusion**


Bubble Sort is an in-place, stable sorting algorithm with $O(1)$ auxiliary space. The optimized version using a swap flag achieves $O(n)$ best-case performance on already-sorted data. However, $O(n^2)$ average and worst-case complexity makes it impractical for large datasets. It is mainly used for educational purposes and small or nearly-sorted arrays.
