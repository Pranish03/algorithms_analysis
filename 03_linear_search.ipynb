# **Lab Report 3: Linear Search**


## **Theory**


Linear Search, also called Sequential Search, is the simplest search algorithm. It scans each element of a list one by one from the beginning until the target element is found or the entire list is exhausted. It requires no pre-sorting of data, making it versatile but less efficient than binary search for large, sorted datasets.

Linear search is fundamental in computer science and is the only viable option when the data is unsorted or when random access is not available (e.g., linked lists). Despite its simplicity, it is widely used for small datasets and as a sub-procedure in more complex algorithms.

## **Algorithm (Pseudocode)**


```
ALGORITHM LinearSearch(arr, n, target)
    Input: Array arr of n elements, search key target
    Output: Index of target if found, else -1

    FOR i <- 0 TO n-1 DO
        IF arr[i] == target THEN
            RETURN i
        END IF
    END FOR
    RETURN -1
```


## **Code & Output**

````python
def linear_search(arr, target):
    for i, val in enumerate(arr):
        if val == target:
            return i
        
    return -1


if __name__ == "__main__":
    arr = [15, 23, 4, 7, 42, 8, 16, 99, 3, 55]
    print(f"Array: {arr}")

    tests = [42, 99, 3, 100]
    for t in tests:
        index = linear_search(arr, t)
        
        if index != -1:
            print(f"Search {t:>3}: FOUND at index {index}")
        else:
            print(f"Search {t:>3}: NOT FOUND")
````

## **Complexity Analysis**


#### **Time Complexity**


1. **Best Case: $O(1)$** <br>
The best case occurs when the target is at index 0. <br>
Thus, $B(n) = O(1)$

2. **Average Case: $O(n)$** <br>
When the target is in middle on average. <br>
Thus, $A(n) = O(n/2) = O(n)$

3. **Worst Case: $O(n)$** <br>
The worst case occurs when the target is at last position or absent. <br>
Thus, $W(n) = O(n)$


#### **Space Complexity** <br>


Linear search uses only a constant amount of extra memory, regardless of input size $n$. <br>
Thus, $S(n) = O(1)$


## **Conclusion**


Linear search is straightforward and works on any data structure without requiring sorting. Its $O(n)$ worst-case and average-case time complexity make it unsuitable for very large datasets. However, for small arrays, unsorted data, or linked lists where random access is unavailable, it remains the algorithm of choice due to its simplicity and zero preprocessing overhead.
