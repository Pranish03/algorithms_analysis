# **Lab Report 8: Kruskal's Algorithm**


## **Theory**


Kruskal's Algorithm finds the Minimum Spanning Tree (MST) of a connected, weighted, undirected graph. The MST connects all vertices with the minimum total edge weight and no cycles. Kruskal's approach is greedy: it sorts all edges by weight, then iteratively adds the cheapest edge that does not form a cycle, using a Union-Find (Disjoint Set Union) data structure to detect cycles efficiently.

The algorithm continues until $n-1$ edges are added (where $n$ is the number of vertices), at which point the MST is complete. It is particularly effective for sparse graphs and is widely used in network design, clustering, and finding approximate solutions to NP-hard problems.

## **Algorithm (Pseudocode)**


```
ALGORITHM Kruskal(Graph G with V vertices, E edges)
    Output: Minimum Spanning Tree edges

    Sort all edges E by weight (ascending)
    Initialize DSU: parent[i] = i for all i
    mst_edges <- empty list
    mst_cost <- 0

    FOR each edge (u, v, weight) in sorted E DO
        IF FIND(u) != FIND(v) THEN
            UNION(u, v)
            ADD (u, v, weight) to mst_edges
            mst_cost <- mst_cost + weight
        END IF
        IF |mst_edges| == V - 1 THEN BREAK
    END FOR
    RETURN mst_edges, mst_cost
```


## **Code & Output**

````python
class DSU:
    def __init__(self, n):
        self.parent = list(range(n))
        self.rank = [0] * n

    def find(self, x):
        if self.parent[x] != x:
            self.parent[x] = self.find(self.parent[x])

        return self.parent[x]

    def union(self, x, y):
        rx, ry = self.find(x), self.find(y)
        if rx == ry:
            return False

        if self.rank[rx] < self.rank[ry]:
            rx, ry = ry, rx

        self.parent[ry] = rx
        if self.rank[rx] == self.rank[ry]:
            self.rank[rx] += 1

        return True


def kruskal(num_vertices, edges):
    edges.sort(key=lambda e: e[2])
    dsu = DSU(num_vertices)
    mst, cost = [], 0

    for u, v, w in edges:
        if dsu.union(u, v):
            mst.append((u, v, w))
            cost += w

            if len(mst) == num_vertices - 1:
                break

    return mst, cost


if __name__ == "__main__":
    edges = [
        (0, 1, 10),
        (0, 2, 6),
        (0, 3, 5),
        (1, 3, 15),
        (2, 3, 4),
        (1, 4, 12),
        (3, 4, 8),
    ]

    mst, total = kruskal(5, edges)
    print("MST Edges (u, v, weight):", mst)
    print("Total MST Cost:", total)
````

## **Complexity Analysis**


#### **Time Complexity**


Kruskal's algorithm is dominated by the initial edge sorting step. <br>
Thus, $T(E) = O(E\:log\:E)$


#### **Space Complexity** <br>


Kruskal’s algorithm mainly uses extra space for storing edges and Disjoint Set (Union-Find) structures.<br>
Thus, $S(V, E) = O(V+E)$


## **Conclusion**


Kruskal's Algorithm efficiently finds the Minimum Spanning Tree in $O(E\:log\:E)$ time, dominated by the initial edge sorting. With path-compressed Union-Find, cycle detection is nearly $O(1)$ per operation. It is ideal for sparse graphs and is widely used in telecommunications network design, approximate clustering, and infrastructure planning problems.
