# **Lab Report 11: N Queens Problem**


## **Theory**


The N-Queens Problem asks: how can $N$ chess queens be placed on an $N×N$ chessboard such that no two queens threaten each other? Queens attack along rows, columns, and diagonals. The Backtracking approach places queens column by column, checking safety at each step and backtracking when no safe position is found.

Backtracking systematically explores the solution space by pruning branches that cannot lead to a valid solution (constraint propagation). For $N=8$, there are $92$ distinct solutions. The N-Queens problem is a canonical example of constraint satisfaction and backtracking, with applications in scheduling, resource allocation, and parallel computing.

## **Algorithm (Pseudocode)**


```
ALGORITHM NQueens(board, col, n)
    Input: board state, current col, board size n
    Output: All valid queen placements

    IF col == n THEN
        PRINT solution
        RETURN
    END IF
    FOR row <- 0 TO n - 1 DO
        IF isSafe(board, row, col) THEN
        board[row][col] <- 1
        NQueens(board, col+1, n)
        board[row][col] <- 0
        END IF
    END FOR

ALGORITHM isSafe(board, row, col)
    Check left in same row
    Check upper-left diagonal
    Check lower-left diagonal
    RETURN TRUE if no conflict
```


## **Code & Output**

````python
def solve_n_queens(n):
    solutions = []
    queens = []

    def is_safe(row, col):
        for c, r in enumerate(queens):
            if r == row or abs(r - row) == abs(c - col):
                return False
        
        return True

    def backtrack(col):
        if col == n:
            solutions.append(queens[:])
            return
        
        for row in range(n):
            if is_safe(row, col):
                queens.append(row)
                backtrack(col + 1)
                queens.pop()

    backtrack(0)
    return solutions


if __name__ == "__main__":
    for size in [4, 6, 8]:
        sols = solve_n_queens(size)
        print(f'N={size}: {len(sols)} solutions found')
    
        if size == 4:
            print(f'  First solution (queen row per col): {sols[0]}')
            board = [['.' if c != sols[0][col] else 'Q'
                    for col in range(size)] for c in range(size)]
            for row in board: print('  ' + ' '.join(row))
````

## **Complexity Analysis**


#### **Time Complexity**


At each row, we try placing a queen in up to $N$ columns, and backtracking prunes invalid placements but in worst case explores factorial permutations of column placements. <br>
Thus, $T(n) = O(n!)$


#### **Space Complexity** <br>


We use recursion depth of $N$ levels (one per row) plus arrays to track columns and diagonals, all of size $O(N)$. <br>
Thus, $S(n) = O(n)$


## **Conclusion**


The Backtracking approach for the N-Queens Problem effectively prunes invalid branches, making it far more efficient than brute-force enumeration. While the theoretical worst case is $O(n!)$, constraint propagation (checking diagonals and rows) reduces the search space dramatically. The space complexity is $O(n)$ for the recursion stack. N-Queens is a foundational problem in AI constraint satisfaction and demonstrates backtracking's power.
