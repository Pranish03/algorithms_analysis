# **Lab Report 7: Merge Sort**


## **Theory**


Merge Sort is a classic divide-and-conquer sorting algorithm. It recursively divides the array into two halves, sorts each half, and then merges the two sorted halves back into a single sorted array. The merge step is the heart of the algorithm — it compares elements from both halves and builds the sorted result.

Merge Sort guarantees $O(n\:log\:n)$ performance in all cases, making it more predictable than Quick Sort. It is a stable sort, preserving the relative order of equal elements, and is the preferred algorithm for sorting linked lists and for external sorting (sorting data too large to fit in memory).


## **Algorithm (Pseudocode)**


```
ALGORITHM MergeSort(arr)
    Input: Array arr
    Output: Sorted array

    IF length(arr) <= 1 THEN RETURN arr
    mid   <- length(arr) / 2
    left  <- MergeSort(arr[0..mid-1])
    right <- MergeSort(arr[mid..end])
    RETURN MERGE(left, right)

ALGORITHM MERGE(left, right)
    result <- empty array
    i <- 0, j <- 0
    WHILE i < length(left) AND j < length(right) DO
        IF left[i] <= right[j] THEN
            APPEND left[i] to result; i++
        ELSE
            APPEND right[j] to result; j++
        END IF
    END WHILE
    APPEND remaining elements of left and right
    RETURN result
```


## **Code & Output**

````python
def merge_sort(arr):
    if len(arr) <= 1:
        return arr
    
    mid = len(arr) // 2
    left  = merge_sort(arr[:mid])
    right = merge_sort(arr[mid:])
    
    return merge(left, right)

def merge(left, right):
    result = []
    i = j = 0
    
    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            result.append(left[i]); i += 1
        else:
            result.append(right[j]); j += 1
    
    result.extend(left[i:])
    result.extend(right[j:])


if __name__ == "__main__":
    datasets = {
        "Random Array":   [6, 3, 1, 7, 4, 2, 5],
        "Sorted Array":   [1, 2, 3, 4, 5, 6, 7],
        "Reversed Array": [7, 6, 5, 4, 3, 2, 1]
    }

    for name, data in datasets.items():
        result = merge_sort(data[:])
        print(f"{name:<15}: {data} -> {result}")
````

## **Complexity Analysis**


#### **Time Complexity**


The merge sort is consistent for all inputs. <br>
Thus, $T(n) = O(n\:log\:n)$


#### **Space Complexity** <br>


Merge sort uses extra memory that grows with input size $n$. <br>
Thus, $S(n) = O(n)$


## **Conclusion**


Merge Sort offers guaranteed $O(n\:log\:n)$ performance across all input types — best, average, and worst case. Its stable nature makes it ideal for sorting data where equal element order must be preserved. The $O(n)$ auxiliary space requirement is its primary drawback compared to in-place algorithms. It is widely used in external sorting and as the base algorithm in Python's Timsort.
