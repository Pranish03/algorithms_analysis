# **Lab Report 5: Insertion Sort**


## **Theory**


Insertion Sort builds the final sorted array one element at a time. Starting from the second element, it compares each element with those in the already-sorted portion of the array and inserts it into its correct position by shifting larger elements one position to the right.

Insertion Sort is efficient for small datasets and nearly-sorted data. It is an in-place, stable, and adaptive sorting algorithm. It is used as the base case in hybrid sorting algorithms such as Timsort (Python's built-in sort) and Introsort (C++ STL sort).

## **Algorithm (Pseudocode)**


```
ALGORITHM InsertionSort(arr, n)
    Input: Array arr of n elements
    Output: Sorted array in ascending order

    FOR i <- 1 TO n-1 DO
        key <- arr[i]
        j   <- i - 1
        WHILE j >= 0 AND arr[j] > key DO
            arr[j+1] <- arr[j]
            j <- j - 1
        END WHILE
        arr[j+1] <- key
    END FOR
    RETURN arr
```


## **Code & Output**

````python
def insertion_sort(arr, n):
    for i in range(1, n):
        key = arr[i]
        j = i - 1

        while j >= 0 and arr[j] > key:
            arr[j + 1] = arr[j]
            j -= 1
            
        arr[j + 1] = key

    return arr


if __name__ == "__main__":
    datasets = {
        "Random Array":   [6, 3, 1, 7, 4, 2, 5],
        "Sorted Array":   [1, 2, 3, 4, 5, 6, 7],
        "Reversed Array": [7, 6, 5, 4, 3, 2, 1]
    }

    for name, data in datasets.items():
        result = insertion_sort(data[:], len(data))
        print(f"{name:<15}: {data} -> {result}")
````

## **Complexity Analysis**


#### **Time Complexity**


1. **Best Case: $O(n)$** <br>
The best case occurs when the array is already sorted (with early-exit flag). <br>
Thus, $B(n) = O(n)$

2. **Average Case: $O(n^2)$** <br>
The average case is when the items in the array is in random order. <br>
Thus, $A(n) = O(n^2)$

3. **Worst Case: $O(n^2)$** <br>
The worst case occurs when the array is reversly sorted. <br>
Thus, $W(n) = O(n^2)$


#### **Space Complexity** <br>


This algorithm does not use any extra memory that grows with input size $n$. <br>
Thus, $S(n) = O(1)$


## **Conclusion**


Insertion Sort is an adaptive, stable, in-place sorting algorithm. Its $O(n)$ best-case performance on nearly-sorted data makes it exceptionally efficient in those scenarios. While $O(n^2)$ worst-case limits its use for large random datasets, its low overhead means it outperforms merge sort and quicksort for small arrays (typically $n < 20$), which is why hybrid algorithms use it for small sub-arrays.
