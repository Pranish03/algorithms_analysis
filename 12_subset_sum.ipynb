# **12. Subset Sum Problem**


## **12.1. Theory**


The Subset Sum Problem asks: given a set of non-negative integers and a target sum, determine whether a subset of the integers exists that adds up to exactly the target sum, and find all such subsets. This is a classic NP-complete decision problem.

The Backtracking approach explores all possible subsets recursively, including or excluding each element and pruning branches where the running sum already exceeds the target. This avoids examining all $2^n$ subsets in many cases. The problem has important applications in cryptography (subset-sum hard problem), resource allocation, and scheduling.

## **12.2. Algorithm (Pseudocode)**


```
ALGORITHM SubsetSum(arr, index, current_sum, target, subset)
    Input: Array arr, current index, running sum, target, current subset
    Output: Print all subsets that sum to target

    IF current_sum == target THEN
        PRINT subset
        RETURN
    END IF
    IF index == n OR current_sum > target THEN
        RETURN
    END IF
    SubsetSum(arr, index + 1, current_sum+arr[index], target, subset + [arr[index]])
    SubsetSum(arr, index + 1, current_sum, target, subset)
```


## **12.3. Code & Output**

````python
def subset_sum(arr, target):
    results = []
    arr.sort()

    def backtrack(index, current_sum, current_subset):
        if current_sum == target:
            results.append(current_subset[:])
            return
        if index >= len(arr) or current_sum > target:
            return

        for i in range(index, len(arr)):
            if current_sum + arr[i] > target:
                break

            current_subset.append(arr[i])
            backtrack(i + 1, current_sum + arr[i], current_subset)
            current_subset.pop()

    backtrack(0, 0, [])
    return results


if __name__ == "__main__":
    test_cases = [
        ([3, 1, 4, 2, 2, 5, 6], 8),
        ([1, 2, 3, 4, 5], 9),
        ([10, 20, 30], 30),
    ]

    for arr, target in test_cases:
        results = subset_sum(arr[:], target)
        print(f"Array: {sorted(arr)}, Target: {target}")
        print(f"  Subsets found ({len(results)}): {results}")
````

## **12.4. Complexity Analysis**


#### **12.4.1. Time Complexity**


At each element, we have two choices (include or exclude), so the recursion forms a binary tree with $2^n$ possible subsets in the worst case. <br>
Thus, $T(n) = O(2^n)$


#### **12.4.2. Space Complexity** <br>


The recursion depth goes up to $n$ levels, and we use a call stack plus a temporary subset array of size at most $n$.<br>
Thus, $S(n) = O(n)$


## **12.5. Conclusion**


The Backtracking approach for Subset Sum with sorting-based pruning significantly reduces the search space compared to the naive $O(2^n)$ approach. Sorting enables early termination when the remaining sum can no longer be achieved. While still exponential in the worst case, it is practical for moderate n. For large inputs, Dynamic Programming with O(n*target) complexity is preferred over backtracking.
