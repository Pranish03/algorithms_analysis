# **9. Dijkastra's Algorithm**


## **9.1. Theory**


Dijkstra's Algorithm finds the shortest path from a single source vertex to all other vertices in a weighted graph with non-negative edge weights. It uses a greedy strategy: it always selects the unvisited vertex with the smallest known distance and relaxes its outgoing edges.

The algorithm maintains a priority queue (min-heap) to efficiently extract the minimum distance vertex. It is widely used in GPS navigation systems, network routing protocols (OSPF), and social network analysis. The algorithm was conceived by Edsger W. Dijkstra in 1956.

## **9.2. Algorithm (Pseudocode)**


```
ALGORITHM Dijkstra(Graph G, source s)
    Input: Weighted graph G, source vertex s
    Output: Shortest distances from s to all vertices

    dist[v] <- INFINITY for all v
    dist[s] <- 0
    priority_queue PQ <- {(0, s)}
    visited <- empty set

    WHILE PQ is not empty DO
        (d, u) <- EXTRACT_MIN(PQ)
        IF u in visited THEN CONTINUE
        ADD u to visited
        FOR each neighbor v of u with edge weight w DO
            IF dist[u] + w < dist[v] THEN
                dist[v] <- dist[u] + w
                INSERT (dist[v], v) into PQ
            END IF
        END FOR
    END WHILE
    RETURN dist
```


## **9.3. Code & Output**

````python
import heapq


def dijkstra(graph, source):
    dist = {v: float("inf") for v in graph}
    dist[source] = 0
    pq = [(0, source)]
    visited = set()

    while pq:
        d, u = heapq.heappop(pq)
        if u in visited:
            continue

        visited.add(u)
        for v, w in graph[u]:
            if dist[u] + w < dist[v]:
                dist[v] = dist[u] + w
                heapq.heappush(pq, (dist[v], v))

    return dist


if __name__ == "__main__":
    graph = {
        0: [(1, 4), (2, 1)], 
        1: [(3, 1)], 
        2: [(1, 2), (3, 5)], 
        3: []
    }

    distances = dijkstra(graph, 0)
    print("Shortest distances from vertex 0:")
    for vertex, dist in sorted(distances.items()):
        print(f"  To vertex {vertex}: {dist}")
````

## **9.4. Complexity Analysis**


#### **9.4.1. Time Complexity**


1. **Using priority queue / min-heap** <br>
Each vertex and edge is processed with heap operations costing $log\:V$. <br>
Thus, $T(V, E) = O((V + E)\:log\:V)$

2. **Using simple array / adjacency matrix** <br>
Each step scans all vertices to find the minimum distance. <br>
Thus, $T(V) = O(V^2)$


#### **9.4.2. Space Complexity** <br>


Dijkastra’s algorithm mainly uses extra space for storing adjacency list ($E$ edges) and distance + visited arrays ($V$ vertices).<br>
Thus, $S(V, E) = O(V+E)$


## **9.5. Conclusion**


Dijkstra's Algorithm with a min-heap priority queue achieves $O((V+E)\:log\:V)$ time complexity, making it efficient for sparse graphs. It is the standard algorithm for shortest-path computation in GPS systems, internet routing (OSPF protocol), and social network analysis. Its limitation is that it only works with non-negative edge weights; for negative weights, Bellman-Ford algorithm must be used.
