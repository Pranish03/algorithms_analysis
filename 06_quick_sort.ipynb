# **Lab Report 5: Quick Sort**


## **Theory**


Quick Sort is a highly efficient, divide-and-conquer sorting algorithm. It selects a 'pivot' element and partitions the array into two sub-arrays: elements smaller than the pivot (left partition) and elements greater than the pivot (right partition). It then recursively sorts each partition.

The choice of pivot is critical to performance. Common strategies include picking the last element (simple but vulnerable to worst case), the median-of-three, or a random pivot. Quick Sort is typically the fastest practical sorting algorithm in practice due to good cache performance and low average-case complexity.

## **Algorithm (Pseudocode)**


```
ALGORITHM QuickSort(arr, low, high)
    Input: Array arr, indices low and high
    Output: Sorted sub-array in place

    IF low < high THEN
        pi <- PARTITION(arr, low, high)
        QuickSort(arr, low, pi - 1)
        QuickSort(arr, pi + 1, high)
    END IF

ALGORITHM PARTITION(arr, low, high)
    pivot <- arr[high]
    i <- low - 1
    FOR j <- low TO high-1 DO
        IF arr[j] <= pivot THEN
            i <- i + 1
            SWAP arr[i] and arr[j]
        END IF
    END FOR
    SWAP arr[i+1] and arr[high]
    RETURN i + 1
```


## **Code & Output**

````python
def partition(arr, low, high):
    pivot = arr[high]
    i = low - 1
    
    for j in range(low, high):
        if arr[j] <= pivot:
            i += 1
            arr[i], arr[j] = arr[j], arr[i]

    arr[i + 1], arr[high] = arr[high], arr[i + 1]
    
    return i + 1

def quick_sort(arr, low=0, high=None):
    if high is None:
        high = len(arr) - 1

    if low < high:
        pi = partition(arr, low, high)
        quick_sort(arr, low, pi - 1)
        quick_sort(arr, pi + 1, high)

    return arr


if __name__ == "__main__":
    datasets = {
        "Random Array":   [64, 34, 25, 12, 22, 11, 90],
        "Sorted Array":   [1, 2, 3, 4, 5, 6, 7],
        "Reversed Array": [7, 6, 5, 4, 3, 2, 1]
    }

    for name, data in datasets.items():
        print(f"{name}: {data} -> {quick_sort(data)}")  
````

## **Complexity Analysis**


#### **Time Complexity**


1. **Best Case: $O(n)$** <br>
The best case occurs when the array is already sorted (with early-exit flag). <br>
Thus, $B(n) = O(n)$

2. **Average Case: $O(n^2)$** <br>
The average case is when the items in the array is in random order. <br>
Thus, $A(n) = O(n^2)$

3. **Worst Case: $O(n^2)$** <br>
The worst case occurs when the array is reversly sorted. <br>
Thus, $W(n) = O(n^2)$


#### **Space Complexity** <br>


This algorithm does not use any extra memory that grows with input size $n$. <br>
Thus, $S(n) = O(1)$
